### 1. What is a Class?
A class is a blueprint or template used to create objects. It defines the data (attributes) and behavior (methods) that objects of that class will have.

Think of a class as an architectural blueprint and objects as actual houses built from it.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(type(student1))
print(type(student2))
```

### 2. What is an Object?
An object is an instance of a class.

Each object has its own data stored in memory.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
```
