### 1. What is a Class?
A class is a blueprint or template used to create objects. It defines the data (attributes) and behavior (methods) that objects of that class will have.

Think of a class as an architectural blueprint and objects as actual houses built from it.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(type(student1))
print(type(student2))
```

### 2. What is an Object?
An object is an instance of a class.

Each object has its own data stored in memory.

```python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
```

### 3.Attributes

Attributes are variables that belong to an object or a class.

They store the object's information.

Example

Student

Name
Age
CGPA

### Instance Attributes

Inside (_init__)and unique for every object

Every Object has its own copy.

```python
class Student: #class definition

    def __init__(self, name, age): #Constructor method
        self.name = name  #attribute
        self.age = age


s1 = Student("Sara", 22) #Object creation
s2 = Student("Ali", 20)

print(s1.name)
print(s2.name) #printing the name attribute of s2 object
```

### Class Attributes

Shared among all Objects.Declared outside Init

```python
class Student:

    university = "University of Layyah"

    def __init__(self, name):
        self.name = name


s1 = Student("Sara")
s2 = Student("Ali")

print(s1.university)
print(s2.university)
```

### 4.Methods 

Methods are functions defined inside a class.

They describe what an object can do.

```python
class Student:

    def study(self):
        print("Studying Python")

    def sleep(self):
        print("Sleeping")


student = Student()

student.study()
student.sleep()
```

### 5. Constructor (__init__)

A constructor is a special method that is automatically executed when an object is created.

```python
class Student:

    def __init__(self):
        print("Constructor Executed")


student = Student()
```

### Constructor with parameters

```python
class Student:

    def __init__(self, name, age):

        self.name = name
        self.age = age


student = Student("Sara", 22)

print(student.name)
print(student.age)
```

### 6.Self keyword

self refers to the current object.

Python automatically passes the current object as the first argument when calling an instance method.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def display(self):
        print(self.name)


student = Student("Sara")
student=Student("Ali")

student.display()
```

### 7.Object Lifecycle

Class Defined

↓

Object Created

↓

Memory Allocated

↓

Constructor Called

↓

Object Used

↓

Object Destroyed

↓

Memory Released

```python
class Student:

    def __init__(self):
        print("Object Created")

    def __del__(self):
        print("Object Destroyed")


student = Student()

del student
```

Note: The timing of __del__() is not guaranteed. Python's garbage collector decides when an object is actually destroyed.

```python
class Student:

    # Class Attribute
    university = "University of Layyah"

    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Method
    def introduce(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
        print(f"My university is {Student.university}")

    # Method
    def birthday(self):
        self.age += 1
        print(f"Happy Birthday! Age is now {self.age}")


# Creating Objects
student1 = Student("Sara", 22)
student2 = Student("Ali", 20)

# Accessing Attributes
print(student1.name)
print(student2.age)

# Calling Methods
student1.introduce()
student2.introduce()

student1.birthday()
student1.introduce()

print(Student.university)
```

1.Employee

**Attributes**: name, salary, department
**Methods**: display_info(), increase_salary(percent)

```python
class Employee:
    def __init__(self,name,salary,department):
        self.name=name
        self.salary=salary
        self.department=department
    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Department: {self.department}")     
    def increase_salary(self,amount):
        self.salary+=amount
        print(f"Salary increased by {amount}. New salary is {self.salary}")
Employee1=Employee("Sara",50000,"IT")
Employee1.display_info()
Employee1.increase_salary(5000)
```

2.Laptop

Attributes: brand, ram, price
Methods: show_specs(), upgrade_ram(new_ram)

```python
class Laptop:
    def __init__(self, brand,ram, price):
        self.brand=brand
        self.ram=ram
        self.price=price
    def show_specs(self):
        print(f"Brand: {self.brand}")
        print(f"RAM: {self.ram} GB")
        print(f"Price: ${self.price}")
    def show_upgrade_options(self,new_ram):
        if new_ram>self.ram:
            print(f"Upgrading RAM from {self.ram} GB to {new_ram} GB")
            self.ram=new_ram
        else:
            print("New RAM must be greater than current RAM")
laptop1=Laptop("Dell",8,800)
laptop1.show_specs()
laptop1.show_upgrade_options(16)
        
```
