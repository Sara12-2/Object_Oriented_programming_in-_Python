### SOLID Principles

SOLID is a set of 5 Object-Oriented Design Principles that help you write clean, maintainable, scalable, and extensible code.

These principles are frequently asked in Python, Java, C#, Backend, and System Design interviews.

S → Single Responsibility Principle

O → Open/Closed Principle

L → Liskov Substitution Principle

I → Interface Segregation Principle

D → Dependency Inversion Principle
