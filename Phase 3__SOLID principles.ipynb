### SOLID Principles

SOLID is a set of 5 Object-Oriented Design Principles that help you write clean, maintainable, scalable, and extensible code.

These principles are frequently asked in Python, Java, C#, Backend, and System Design interviews.

S → Single Responsibility Principle

O → Open/Closed Principle

L → Liskov Substitution Principle

I → Interface Segregation Principle

D → Dependency Inversion Principle

### S — Single Responsibility Principle (SRP)
A class should have only one reason to change.

Meaning:

A class should perform only one job.

```python
# class Student:

#     def calculate_grade(self):
#         print("Calculating Grade")

#     def save_to_database(self):
#         print("Saving to Database")

#     def send_email(self):
#         print("Sending Email")
```

```python
class Student:

    def calculate_grade(self):
        print("Calculating Grade")


class StudentRepository:

    def save(self):
        print("Saving Student")


class EmailService:

    def send(self):
        print("Sending Email")
```

### O — Open/Closed Principle (OCP)
Software entities should be open for extension but closed for modification.

Meaning:

You should add new functionality without changing existing code.

```python
# class Payment:

#     def pay(self, method):

#         if method == "CreditCard":
#             print("Credit Card")

#         elif method == "PayPal":
#             print("PayPal")
```

```python
class Payment:

    def pay(self):
        pass


class CreditCard(Payment):

    def pay(self):
        print("Credit Card")


class PayPal(Payment):

    def pay(self):
        print("PayPal")


class Crypto(Payment):

    def pay(self):
        print("Crypto")
```

### L — Liskov Substitution Principle (LSP)

A child class should be able to replace its parent class without breaking the program.

If code expects a parent object, any child object should work correctly.

```python
class Bird:

    def move(self):
        print("Bird is moving")


class Sparrow(Bird):

    def move(self):
        print("Flying")


class Penguin(Bird):

    def move(self):
        print("Walking")
```

### I — Interface Segregation Principle (ISP)
Clients should not be forced to depend on methods they do not use.

In Python, this usually means creating small abstract base classes instead of one large interface.
