### 1. Encapsulation
Encapsulation is the process of bundling data (attributes) and methods together inside a class while controlling access to the data.

#### With encapsulation:

Validation is added.
Data remains safe.
Objects stay in a valid state.

Encapsulation = Data + Methods + Access Control

Unlike Java and C++, Python doesn't enforce access modifiers. Instead, it uses naming conventions and name mangling.

```python
class BankAccount:

    def __init__(self):

        self.__balance = 5000

    def deposit(self, amount):

        self.__balance += amount

    def withdraw(self, amount):

        if amount <= self.__balance:
            self.__balance -= amount

    def get_balance(self):

        return self.__balance


account = BankAccount()

account.deposit(2000)

account.withdraw(1000)

print(account.get_balance())
```

### Setter

Updates data with validation.

```python
class Student:

    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):

        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid Marks")


student = Student()

student.set_marks(95)

```

### Modern Python Way (@property)

```python
class Student:

    def __init__(self):
        self._marks = 0

    @property
    def marks(self):
        return self._marks

    @marks.setter
    def marks(self, value):

        if 0 <= value <= 100:
            self._marks = value
        else:
            print("Invalid Marks")


student = Student()

student.marks = 85

print(student.marks)
```

### Instance Method

### Class Method 

A class method is a method that works with the class itself instead of a specific object.

```python
class Book:
    pages=100
    @classmethod 
    def read(cls):
        print(f"Reading {cls.pages} pages")
Book.read()
```

### Static method 

A static method is a method that belongs to a class but does not depend on either the object (self) or the class (cls).

```python
class Math:

    @staticmethod
    def add(a, b):
        return a + b


print(Math.add(10, 20))
```

## Inheritance 

Inheritance is an OOP principle that allows one class (child class) to acquire the properties and methods of another class (parent class).

It promotes code reusability, extensibility, and hierarchical relationships.

Parent Class also called Base and super class and child class also called derived or sub class 

```python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Woof")


dog = Dog()

dog.eat()

dog.bark()
```

```python
class Vehicle:

    def start(self):
        print("Vehicle Started")


class Car(Vehicle):

    def drive(self):
        print("Driving")


car = Car()

car.start()

car.drive()
```

### Constructor Inheritance

The child inherits the parent's constructor if it doesn't define its own.

```python
class Animal:

    def __init__(self):

        print("Animal Constructor")


class Dog(Animal):

    pass


dog = Dog()
```

### Child Constructor

```python
class Animal:

    def __init__(self, name):

        self.name = name


class Dog(Animal):

    def __init__(self, name, breed):

        super().__init__(name) #Inherit name of parent by using Super()

        self.breed = breed


dog = Dog("Buddy", "Labrador")

print(dog.name)

print(dog.breed)
```

### Method Overriding

Child replaces the parent's implementation.

```python
class Animal:

    def sound(self):
        print("Animal Sound")


class Dog(Animal):

    def sound(self):
        print("Woof") # Prefer itself than parent 


dog = Dog()

dog.sound()
```

### Multilevel Inheritance

A TO B TO C

```python
class Animal:

    def eat(self):
        print("Eating")


class Mammal(Animal):

    def walk(self):
        print("Walking")


class Dog(Mammal):

    def bark(self):
        print("Woof")


dog = Dog()

dog.eat()

dog.walk()

dog.bark()
```

### Polymorphism

Polymorphism means "many forms." It allows different objects to respond to the same method call in their own way.

One interface, multiple implementations.

The same method name behaves differently depending on the object calling it.

```python
class Dog:

    def sound(self):
        print("Woof")


class cat:

    def sound(self):
        print("Meow")


class Cow:

    def sound(self):
        print("Moo")


animals = [Dog(), cat(), Cow()]

for animal in animals:
    animal.sound()
```

### Method Overriding (Runtime Polymorphism)

```python
class Animal:

    def sound(self):
        print("Animal Sound")
class Dog(Animal):

    def sound(self):
        print("Woof")
dog = Dog()

dog.sound()
```

The child overrides the parent's method.

```python
class Employee:

    def work(self):
        print("Employee Working")


class Developer(Employee):

    def work(self):
        print("Writing Code")


class Designer(Employee):

    def work(self):
        print("Designing UI")


employees = [Developer(), Designer()]

for employee in employees:
    employee.work()
```

### Built-in Polymorphism

```python
print(len("Python"))

print(len([1,2,3]))

print(len({"a":1}))
```

```python
class AIModel:

    def predict(self):
        pass


class Chatbot(AIModel):

    def predict(self):
        print("Generating Response")


class ImageClassifier(AIModel):

    def predict(self):
        print("Classifying Image")


class Translator(AIModel):

    def predict(self):
        print("Translating Text")


models = [
    Chatbot(),
    ImageClassifier(),
    Translator()
]

for model in models:
    model.predict()
```
